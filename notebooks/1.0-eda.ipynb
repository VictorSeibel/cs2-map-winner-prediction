```python
# %%
stats = ['kills', 'deaths', 'assists', 'adr', 'kast', 'kddiff']

team_stats = [f'team{team}_player{player}_{stat}'
    for team in [1, 2]
    for player in range(1, 6)
    for stat in stats]

other_stats = ['is_total', 'bestOf', 'score1_game', 'score2_game', 'datetime', 'games_played']

do_not_count_summary = cs2_all_tier_games_df[cs2_all_tier_games_df['is_total'] == False]

do_not_count_summary[(team_stats + other_stats)].describe()
```

```python
# %%

# not missing values

def missing_values_columns(df):
    missing_values_columns = (df.isna().sum() > 0)
    missing_values_columns_df = pd.DataFrame(missing_values_columns, columns= ['result']).reset_index(names= 'match_columns')
    return [match_columns for match_columns, result in zip(missing_values_columns_df['match_columns'], missing_values_columns_df['result']) if result]
    
```

```python
# %%

# concat team1_id and team2_id columns
teams = pd.concat([
    cs2_all_tier_games_df[['tournament', 'team1_id']].rename(columns={'team1_id': 'team_id'}),
    cs2_all_tier_games_df[['tournament', 'team2_id']].rename(columns={'team2_id': 'team_id'})
], ignore_index=True)

# Verifying if for each tournament i have a unique team_id
is_team_id_unique = teams.groupby('team_id')['tournament'].nunique().eq(1).all()

print(is_team_id_unique)
```

```python
# %%

# concat team1_id and team2_id columns
team_player = pd.concat([
    cs2_all_tier_games_df[
        ['team1_id',
         'team1_player1_id', 'team1_player2_id', 'team1_player3_id',
         'team1_player4_id', 'team1_player5_id']
    ].rename(columns={
        'team1_id': 'team_id',
        'team1_player1_id': 'player1_id',
        'team1_player2_id': 'player2_id',
        'team1_player3_id': 'player3_id',
        'team1_player4_id': 'player4_id',
        'team1_player5_id': 'player5_id'
    }),

    cs2_all_tier_games_df[
        ['team2_id',
         'team2_player1_id', 'team2_player2_id', 'team2_player3_id',
         'team2_player4_id', 'team2_player5_id']
    ].rename(columns={
        'team2_id': 'team_id',
        'team2_player1_id': 'player1_id',
        'team2_player2_id': 'player2_id',
        'team2_player3_id': 'player3_id',
        'team2_player4_id': 'player4_id',
        'team2_player5_id': 'player5_id'
    })
])

is_player_unique = [
    team_player.groupby('team_id')[f'player{i}_id'].nunique(dropna=False).eq(1).all()
    for i in range(1, 6)
]

print(is_player_unique)
```
